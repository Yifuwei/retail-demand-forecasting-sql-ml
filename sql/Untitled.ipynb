```python
import sqlite3
import pandas as pd
conn = sqlite3.connect("retail.db")
print("Database connected")
```

```python
conn.executescript("""
DROP TABLE IF EXISTS sales;

CREATE TABLE sales (
    item_id TEXT,
    store_id TEXT,
    units_sold INTEGER
);

INSERT INTO sales VALUES
('FOODS_001', 'CA_1', 12),
('FOODS_002', 'CA_1', 8),
('FOODS_003', 'CA_1', 20),
('FOODS_001', 'CA_2', 15),
('FOODS_002', 'CA_2', 5);
""")

conn.commit()
```

```python
query = """
SELECT
    item_id,
    store_id,
    units_sold
FROM sales
WHERE units_sold > 10
ORDER BY units_sold DESC;
"""

df = pd.read_sql_query(query, conn)

display(df)
```

```python
query = """ SELECT * FROM sales limit 50; """
df = pd.read_sql_query(query, conn)
display(df)
```

```python
query = """ SELECT distinct item_id FROM sales where store_id="CA_1"; """
df = pd.read_sql_query(query, conn)
display(df)
```

```python
query = """ SELECT COUNT(item_id) FROM sales where store_id="CA_1"; """
df = pd.read_sql_query(query, conn)
display(df)
```
